```python
import numpy as np
import matplotlib.pyplot as plt
import os
import pickle
import matplotlib.colors as mcolors
import matplotlib.cm as cm
```

```python
###########################################################

notebook_dir = os.getcwd()

os.chdir(notebook_dir)

###########################################################
```

```python
# Load uniform MPS data
try:
    with open('Fig_8_uMPS_data.pkl', 'rb') as f:
        Lmax_dict_32, Lmax_dict_64 = pickle.load(f)

except:
    print("Something went wrong")
```

```python
fs = 21
ms = 7.5
lw = 1

plt.switch_backend('pgf')
#
fig, ax = plt.subplots(figsize=(6,4))
#
plt.gcf().subplots_adjust(left=0.15)
plt.gcf().subplots_adjust(bottom=0.175)
#
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": "Times New Roman",
    "font.size": fs,
})
#
################################################################

fig, ax = plt.subplots(figsize=(10.0, 8.0))

tQ = 5.81537
t0 = 0.0370717


filtered_data_32 = [
    (keys, data_list) 
    for keys, data_list in Lmax_dict_32.items()
]
filtered_data_64 = [
    (keys, data_list) 
    for keys, data_list in Lmax_dict_64.items()
]

min_tauQ = min(min(item[0] for item in filtered_data_32), min(item[0] for item in filtered_data_64))
max_tauQ = max(max(item[0] for item in filtered_data_32), max(item[0] for item in filtered_data_64))


norm = mcolors.LogNorm(vmin=min_tauQ, vmax=max_tauQ)

Jf = 26.9748 # 

cmap = plt.get_cmap('viridis')

ax.plot([], [], label=r'$D=32$', c='black', marker='.', ms=ms, lw=lw)
ax.plot([], [], label=r'$D=64$', c='black', marker='+', ms=ms, lw=lw)

xx = np.logspace(np.log10(1e-2), np.log10(1e0))
ax.plot(xx, 0.35/xx, c='red', ls='--', lw=2*lw, label=r'$\propto (h \tau_Q)^{-1}$', zorder=10)

for key, data_list in Lmax_dict_32.items():
    tauQ = key
    hz = data_list[0]
    Lmax = data_list[1]

    color = cmap(norm(tauQ))    
    
    plt.plot(hz * tauQ, Lmax, marker='.', ms=ms, c=color, lw=lw, zorder=5)

for key, data_list in Lmax_dict_64.items():
    tauQ = key
    hz = data_list[0]
    Lmax = data_list[1]

    color = cmap(norm(tauQ))    
    
    plt.plot(hz * tauQ, Lmax, marker='+', ms=ms, c=color, lw=lw, zorder=5)

ax.set_ylim(1, 32)
ax.set_xlim(2 * 10**-3, 4 * 10**-1)
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel(r'$h \tau_{Q}$', fontsize=fs)
ax.set_ylabel(r'$L_\mathrm{max}$', fontsize=fs)

ax.legend(loc='best', ncols=1, frameon=False, fontsize=fs-2)
cbar = fig.colorbar(cm.ScalarMappable(norm=norm, cmap=cmap), ax=ax, orientation="vertical", aspect=50, fraction=0.05, pad=0.025)
cbar.set_label(r'$\tau_Q$', size=fs)

fig.tight_layout(rect=[0, 0, 1, 1], pad=0.1)
plt.savefig('Fig_8.pdf')
plt.close()
```
